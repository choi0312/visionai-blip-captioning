# BLIP Image Captioning + English-to-Korean Translation

Colab에서 GPU 런타임을 선택합니다. 공개 COCO 사진 100장(약 17MB)에 두 디코딩 조건을 적용하고 단어별 교차 Attention과 한국어 번역을 저장합니다. 추가 학습은 없습니다.


```python
from pathlib import Path
import os,subprocess,sys
repo=Path("/content/visionai-blip-captioning")
if not repo.exists():
    subprocess.run(["git","clone","https://github.com/choi0312/visionai-blip-captioning.git",str(repo)],check=True)
os.chdir(repo)
subprocess.run([sys.executable,"-m","pip","install","-r","requirements.txt"],check=True)
os.environ["VISIONAI_ROOT"]=str(repo)
```

각 스크립트를 별도 프로세스로 실행합니다. 설치 전에 이미 import한 패키지의 영향을 줄이며, 재실행은 results/와 figures/를 덮어씁니다. 평가 토크나이저에는 Java가 필요합니다(Colab 제공 환경).


```python
for script in ["download_images.py","caption_inference.py","evaluate_captions.py","analyze_and_plot.py","verify_results.py"]:
    subprocess.run([sys.executable,str(Path("experiments")/script)],check=True)
```

```python
import json
from IPython.display import Image,display
print(json.dumps(json.loads(Path("results/metrics.json").read_text()),ensure_ascii=False,indent=2))
for f in ["caption_examples.png","decoding_comparison.png","cross_attention.png","korean_translation.png"]:
    display(Image(filename=str(Path("figures")/f)))
```
